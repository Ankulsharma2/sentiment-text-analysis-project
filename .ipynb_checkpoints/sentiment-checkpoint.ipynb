```python
# Import libraries
import re
from nltk.stem import PorterStemmer
from nltk.corpus import stopwords
import pandas as pd

# Initialize stemmer
stemmer = PorterStemmer()

# Ensure stopwords are downloaded
import nltk
nltk.download('stopwords')
STOPWORDS = set(stopwords.words('english'))

# Sample data for testing (uncomment if `data` is undefined)
# data = pd.DataFrame({'verified_reviews': ["I love this product!", "It's great and very useful.", "Not bad, but could be better."]})

corpus = []

# Process each review
for i in range(0, data.shape[0]):
    # Extract and clean the review text
    
    review = review.lower().split()
    
    # Stem and remove stopwords
    review = [stemmer.stem(word) for word in review if word not in STOPWORDS]
    review = ' '.join(review)
    
    # Append to corpus
    corpus.append(review)

print(corpus)  # Print corpus to verify the output
```

```python
corpus = []
stemmer = PorterStemmer()
for i in range(0, data.shape[0]):
 
  review = review.lower().split()
  review = [stemmer.stem(word) for word in review if not word in STOPWORDS]
  review = ' '.join(review)
  corpus.append(review)
```

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer(max_features = 2500)

#Storing independent and dependent variables in X and y
X = cv.fit_transform(corpus).toarray()
y = data['feedback'].values
```

```python
#Saving the Count Vectorizer
import pickle
import os

# Create the 'Models' directory if it doesn't exist
if not os.path.exists("Models"):
    os.makedirs("Models")

cv = CountVectorizer(max_features = 2500)

#Storing independent and dependent variables in X and y
X = cv.fit_transform(corpus).toarray()
y = data['feedback'].values


pickle.dump(cv, open('Models/countVectorizer.pkl', 'wb'))
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer

# Sample text data and labels
corpus = [
    "This is the first document.",
    "This is the second document.",
    "And this is the third document.",
    "Is this the first document?"
]
labels = [0, 1, 0, 1]

# Initialize and fit the TF-IDF vectorizer
tfidf = TfidfVectorizer()
X = tfidf.fit_transform(corpus)

# Split the dataset
X_train, X_test, y_train, y_test = train_test_split(X, labels, test_size=0.2, random_state=42)

# Define and train the classifier
clf = RandomForestClassifier()
clf.fit(X_train, y_train)

# Save the classifier and TF-IDF vectorizer
import pickle
pickle.dump(clf, open('clf.pkl', 'wb'))
pickle.dump(tfidf, open('tfidf.pkl', 'wb'))
```

```python
import re
import pickle

# Define preprocessing
def preprocessing(comment):
    comment = comment.lower()
    comment = re.sub(	'Great', comment)
    comment = re.sub(r'\s+', ' ', comment).strip()
    return comment

# Load trained models
clf = pickle.load(open('clf.pkl', 'rb'))
tfidf = pickle.load(open('tfidf.pkl', 'rb'))

# Define prediction function
def prediction(comment):
    preprocessed_comment = preprocessing(comment)
    comment_list = [preprocessed_comment]  # Wrap the preprocessed comment in a list
    comment_vector = tfidf.transform(comment_list)
    pred = clf.predict(comment_vector)[0]
    return pred

# Test prediction
print
```

```python
if prediction == 1:
    print("positive comment")
else:
    print("negative comment")
```
